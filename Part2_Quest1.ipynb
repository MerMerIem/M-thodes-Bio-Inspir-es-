###Question 1: Implémentation de G en utilisant un dictionnaire


```python
# Chaque clé est un sommet, chaque valeur est la liste de ses voisins
# (lue ligne par ligne dans la matrice d'adjacence)
G = {
    'a': ['b', 'c'],
    'b': ['a', 'd', 'e'],
    'c': ['a', 'd'],
    'd': ['b', 'c', 'e'],
    'e': ['b', 'd', 'f', 'g'],
    'f': ['e', 'g'],
    'g': ['e', 'f', 'h'],
    'h': ['g']
}

# Affichage du dictionnaire
print("Graphe G (dictionnaire) :")
for sommet, voisins in G.items():
    print(sommet, "->", voisins)
```

```python
# (Optionnel) Tracé du graphe pour vérifier visuellement
import networkx as nx
import matplotlib.pyplot as plt

graphe = nx.Graph(G)
pos = {'a': (0, 1), 'b': (1, 2), 'c': (1, 0), 'd': (2, 1),
       'e': (3, 2), 'g': (4, 1), 'h': (3, 0), 'f': (5, 2)}
nx.draw(graphe, pos, with_labels=True, node_color='lightblue',
        node_size=800, font_size=14, edge_color='gray')
plt.show()
```
